```python
class VacuumCleaner:

    def __init__(self):
        # Initial state
        self.position = "A"
        self.room_A = "Dirty"
        self.room_B = "Dirty"

    def display(self):
        print("\nCurrent State:")
        print("Vacuum Position :", self.position)
        print("Room A          :", self.room_A)
        print("Room B          :", self.room_B)

    def clean(self):
        if self.position == "A":
            if self.room_A == "Dirty":
                print("Action: SUCK")
                self.room_A = "Clean"
            else:
                print("Room A is already clean.")

        elif self.position == "B":
            if self.room_B == "Dirty":
                print("Action: SUCK")
                self.room_B = "Clean"
            else:
                print("Room B is already clean.")

    def move_right(self):
        if self.position == "A":
            print("Action: MOVE RIGHT")
            self.position = "B"

    def move_left(self):
        if self.position == "B":
            print("Action: MOVE LEFT")
            self.position = "A"

    def goal_reached(self):
        return self.room_A == "Clean" and self.room_B == "Clean"

    def run(self):

        while not self.goal_reached():

            self.display()

            # If current room is dirty, clean it
            if self.position == "A" and self.room_A == "Dirty":
                self.clean()

            elif self.position == "B" and self.room_B == "Dirty":
                self.clean()

            # If current room is clean, move to the other room
            elif self.position == "A":
                self.move_right()

            elif self.position == "B":
                self.move_left()

        self.display()
        print("\nGoal reached!")
        print("Both rooms are clean.")


# Create vacuum cleaner
vacuum = VacuumCleaner()

# Start the agent
vacuum.run()
```
