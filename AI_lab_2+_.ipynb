```python

def ids(start, goal):
    def dls(state, depth, path, visited):
        if state == goal:
            return path

        if depth == 0:
            return None

        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = []
        if row > 0:
            moves.append(zero - 3)
        if row < 2:
            moves.append(zero + 3)
        if col > 0:
            moves.append(zero - 1)
        if col < 2:
            moves.append(zero + 1)

        for pos in moves:
            new_state = list(state)
            new_state[zero], new_state[pos] = new_state[pos], new_state[zero]
            new_state = tuple(new_state)

            if new_state not in visited:
                result = dls(new_state, depth - 1,
                             path + [new_state],
                             visited | {new_state})
                if result is not None:
                    return result
        return None

    depth = 0
    while True:
        result = dls(start, depth, [start], {start})
        if result is not None:
            return result
        depth += 1


start = (5, 4, 0,
         6, 1, 8,
         7, 3, 2)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = ids(start, goal)

print("Solution:")
for state in solution:
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()
```

```python
# 8 Puzzle Problem using DFS

# Goal state
GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

def get_neighbors(state):
    """Generate possible moves."""

    neighbors = []

    zero_pos = state.index(0)
    row = zero_pos // 3
    col = zero_pos % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_pos = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with the selected tile
            new_state[zero_pos], new_state[new_pos] = \
                new_state[new_pos], new_state[zero_pos]

            neighbors.append(tuple(new_state))

    return neighbors


def dfs(start):
    """Depth First Search."""

    stack = [(start, [start])]
    visited = set()

    while stack:

        state, path = stack.pop()

        # If goal is reached
        if state == GOAL:
            return path

        if state in visited:
            continue

        visited.add(state)

        # Add neighboring states to stack
        for next_state in get_neighbors(state):

            if next_state not in visited:
                stack.append(
                    (next_state, path + [next_state])
                )

    return None


def print_puzzle(state):
    """Print puzzle in 3x3 format."""

    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])

    print()


# -------------------------
# Main Program
# -------------------------

start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

print("Initial State:")
print_puzzle(start)

solution = dfs(start)

if solution:

    print("Solution found!")
    print("Number of moves:", len(solution) - 1)

    print("\nSteps:")

    for i, state in enumerate(solution):

        print("Step", i)
        print_puzzle(state)

else:
    print("No solution found.")
```

```python
import heapq

initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Heuristic: Manhattan Distance
def heuristic(state):
    distance = 0

    for i in range(9):

        if state[i] == 0:
            continue

        # Current position
        row = i // 3
        col = i % 3

        # Goal position
        goal_pos = goal.index(state[i])
        goal_row = goal_pos // 3
        goal_col = goal_pos % 3

        distance += abs(row - goal_row)
        distance += abs(col - goal_col)

    return distance


# Generate possible moves
def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


# A* algorithm
def astar():
    pq = []

    heapq.heappush(pq, (heuristic(initial), 0, initial))

    cost = {initial: 0}
    parent = {initial: None}

    while pq:
        f, g, current = heapq.heappop(pq)

        if current == goal:
            break

        for next_state in get_neighbors(current):

            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:

                cost[next_state] = new_g

                h = heuristic(next_state)
                new_f = new_g + h

                parent[next_state] = current
                heapq.heappush(pq, (new_f, new_g, next_state))

    # Construct path
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    print("Solution:")

    for state in path:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()

    print("Number of moves:", len(path) - 1)


astar()
```

```python

# Hill Climbing for N-Queens

# Calculate the cost
def calculate_cost(state):
    cost = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):

            # Same row
            if state[i] == state[j]:
                cost += 1

            # Same diagonal
            elif abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


# Generate neighboring states
def get_neighbors(state):
    neighbors = []
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            new_state = state.copy()

            # Swap two queens
            new_state[i], new_state[j] = new_state[j], new_state[i]

            neighbors.append(new_state)

    return neighbors


# Hill Climbing algorithm
def hill_climbing(state):

    while True:
        current_cost = calculate_cost(state)

        print("\nCurrent State:", state)
        print("Current Cost:", current_cost)

        # Goal reached
        if current_cost == 0:
            print("Solution Found!")
            return state

        neighbors = get_neighbors(state)

        best_state = state
        best_cost = current_cost

        # Find the best neighbor
        for neighbor in neighbors:
            cost = calculate_cost(neighbor)

            print("Neighbor:", neighbor, "Cost:", cost)

            if cost < best_cost:
                best_state = neighbor
                best_cost = cost

        # No better neighbor
        if best_cost >= current_cost:
            print("\nNo better neighbor found.")
            print("Hill Climbing stopped.")
            return state

        state = best_state


# Main program
n = int(input("Enter number of queens: "))

print("Enter the initial state.")
print("Enter", n, "row positions separated by spaces.")

state = list(map(int, input().split()))

if len(state) != n:
    print("Invalid input!")
else:
    solution = hill_climbing(state)

    print("\nFinal State:", solution)
    print("Final Cost:", calculate_cost(solution))
```

```python

# Propositional Logic using Truth Table Enumeration

# Implication: P -> Q is false only when P is True and Q is False
def implies(p, q):
    return (not p) or q


# Check the Knowledge Base
def check_kb(p, q, r):

    s1 = implies(q, p)       # Q -> P
    s2 = implies(p, not q)   # P -> ~Q
    s3 = q and r             # Q ^ R

    return s1 and s2 and s3


# Check the query
def check_query(p, q, r, query):

    if query == "R":
        return r

    elif query == "R->P":
        return implies(r, p)

    elif query == "Q->R":
        return implies(q, r)


# Main program
query = input("Enter query (R / R->P / Q->R): ")

print("\nP\tQ\tR\tKB\tQuery")

entails = True

# Generate all possible truth values
for p in [False, True]:
    for q in [False, True]:
        for r in [False, True]:

            kb = check_kb(p, q, r)
            result = check_query(p, q, r, query)

            print(p, "\t", q, "\t", r, "\t", kb, "\t", result)

            # If KB is true but query is false,
            # then KB does not entail the query
            if kb and not result:
                entails = False


if entails:
    print("\nKB entails", query)
else:
    print("\nKB does not entail", query)
```

```python

import re

# Check whether a term is a variable
def is_variable(x):
    return isinstance(x, str) and x[0].islower()

# Convert input into a predicate structure
def parse(s):
    tokens = re.findall(r'[A-Za-z0-9_]+|[(),]', s)
    pos = [0]

    def read():
        name = tokens[pos[0]]
        pos[0] += 1

        if pos[0] < len(tokens) and tokens[pos[0]] == '(':
            pos[0] += 1
            args = []

            while True:
                args.append(read())

                if tokens[pos[0]] == ')':
                    pos[0] += 1
                    break

                pos[0] += 1

            return (name, args)

        return name

    return read()

# Apply substitutions
def apply(term, subst):
    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    return (term[0], [apply(x, subst) for x in term[1]])

# Check occurs condition
def occurs(var, term, subst):
    term = apply(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, x, subst) for x in term[1])

    return False

# Unification function
def unify(a, b, subst):
    a = apply(a, subst)
    b = apply(b, subst)

    if a == b:
        return True

    if is_variable(a):
        if occurs(a, b, subst):
            return False
        subst[a] = b
        return True

    if is_variable(b):
        if occurs(b, a, subst):
            return False
        subst[b] = a
        return True

    if isinstance(a, tuple) and isinstance(b, tuple):
        if a[0] != b[0] or len(a[1]) != len(b[1]):
            return False

        for x, y in zip(a[1], b[1]):
            if not unify(x, y, subst):
                return False

        return True

    return False

# Main program
e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

a = parse(e1)
b = parse(e2)

subst = {}

if unify(a, b, subst):
    print("Unification Successful!")
    print("Substitutions:")

    for key, value in subst.items():
        print(key, "=", value)

    print("Unified Expression:", apply(a, subst))
else:
    print("Unification Failed!")
```

First Order Logic – Forward Chaining

```python

# Forward Chaining in First Order Logic

# Initial facts
facts = [
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Missile", "T1"),
    ("Owns", "A", "T1")
]

# Rules: (Conditions, Conclusion)
rules = [
    ([("Missile", "x")],
     ("Weapon", "x")),

    ([("Owns", "A", "x"), ("Missile", "x")],
     ("Sells", "Robert", "x", "A")),

    ([("Enemy", "x", "America")],
     ("Hostile", "x")),

    ([("American", "x"), ("Weapon", "y"),
      ("Sells", "x", "y", "z"), ("Hostile", "z")],
     ("Criminal", "x"))
]

# Check whether a term is a variable
def is_variable(x):
    return x[0].islower()

# Match two terms
def match(a, b, subst):
    if is_variable(a):
        if a in subst:
            return subst if subst[a] == b else None
        subst[a] = b
        return subst

    if a == b:
        return subst

    return None

# Match a rule condition with a fact
def match_fact(pattern, fact, subst):
    if len(pattern) != len(fact):
        return None

    for a, b in zip(pattern, fact):
        subst = match(a, b, subst.copy())

        if subst is None:
            return None

    return subst

# Replace variables with their values
def substitute(atom, subst):
    return tuple(subst.get(x, x) for x in atom)

# Forward Chaining
def forward_chaining():
    global facts

    new = True

    while new:
        new = False

        for conditions, conclusion in rules:

            substitutions = [{}]

            for condition in conditions:
                updated = []

                for sub in substitutions:
                    for fact in facts:
                        result = match_fact(condition, fact, sub.copy())

                        if result is not None:
                            updated.append(result)

                substitutions = updated

            for sub in substitutions:
                result = substitute(conclusion, sub)

                if result not in facts:
                    facts.append(result)
                    print("Derived:", result)
                    new = True

    # Check the goal
    if ("Criminal", "Robert") in facts:
        print("\nGoal Proved: Robert is a criminal.")
    else:
        print("\nGoal Cannot Be Proved.")

# Main program
print("Initial Facts:", facts)
print("\nForward Chaining Process:")
forward_chaining()
```

```python

# Resolution in First Order Logic

VARIABLES = {"x", "y", "z", "w"}

# Unification
def unify(a, b, subst):
    while a in subst:
        a = subst[a]

    while b in subst:
        b = subst[b]

    if a == b:
        return subst

    if a in VARIABLES:
        subst[a] = b
        return subst

    if b in VARIABLES:
        subst[b] = a
        return subst

    return None


# Apply substitutions
def apply(literal, subst):
    neg, pred, args = literal

    new_args = []
    for x in args:
        while x in subst:
            x = subst[x]
        new_args.append(x)

    return (neg, pred, tuple(new_args))


# Resolution function
def resolve(c1, c2):

    for i, l1 in enumerate(c1):
        for j, l2 in enumerate(c2):

            if l1[0] != l2[0] and l1[1] == l2[1]:

                if len(l1[2]) != len(l2[2]):
                    continue

                subst = {}
                valid = True

                for a, b in zip(l1[2], l2[2]):
                    result = unify(a, b, subst)

                    if result is None:
                        valid = False
                        break

                if valid:
                    new_clause = []

                    for k, lit in enumerate(c1):
                        if k != i:
                            new_clause.append(apply(lit, subst))

                    for k, lit in enumerate(c2):
                        if k != j:
                            new_clause.append(apply(lit, subst))

                    return new_clause

    return None


# CNF Clauses

# C1: NOT Food(x) OR Likes(John,x)
C1 = [
    (True, "Food", ("x",)),
    (False, "Likes", ("John", "x"))
]

# C2: NOT Eats(x,y) OR Killed(x) OR Food(y)
C2 = [
    (True, "Eats", ("x", "y")),
    (False, "Killed", ("x",)),
    (False, "Food", ("y",))
]

# C3: Eats(Anil,Peanuts)
C3 = [
    (False, "Eats", ("Anil", "Peanuts"))
]

# C4: Alive(Anil)
C4 = [
    (False, "Alive", ("Anil",))
]

# C5: NOT Alive(x) OR NOT Killed(x)
C5 = [
    (True, "Alive", ("x",)),
    (True, "Killed", ("x",))
]

# C6: Negation of goal
C6 = [
    (True, "Likes", ("John", "Peanuts"))
]


# Perform resolution
print("Resolution Proof:\n")

r1 = resolve(C1, C6)
print("Step 1:", r1)

r2 = resolve(C2, C3)
print("Step 2:", r2)

r3 = resolve(C5, C4)
print("Step 3:", r3)

if r1 is not None and r2 is not None and r3 is not None:

    r4 = resolve(r2, r3)
    print("Step 4:", r4)

    if r4 is not None:
        r5 = resolve(r1, r4)
        print("Step 5:", r5)

        if r5 == []:
            print("\nEmpty clause obtained!")
            print("Goal Proved: John likes peanuts.")
        else:
            print("\nGoal Cannot Be Proved.")

    else:
        print("Resolution failed at Step 4.")

else:
    print("Resolution failed in the initial steps.")
```
