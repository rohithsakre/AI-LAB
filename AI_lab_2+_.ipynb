```python

def ids(start, goal):
    def dls(state, depth, path, visited):
        if state == goal:
            return path

        if depth == 0:
            return None

        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = []
        if row > 0:
            moves.append(zero - 3)
        if row < 2:
            moves.append(zero + 3)
        if col > 0:
            moves.append(zero - 1)
        if col < 2:
            moves.append(zero + 1)

        for pos in moves:
            new_state = list(state)
            new_state[zero], new_state[pos] = new_state[pos], new_state[zero]
            new_state = tuple(new_state)

            if new_state not in visited:
                result = dls(new_state, depth - 1,
                             path + [new_state],
                             visited | {new_state})
                if result is not None:
                    return result
        return None

    depth = 0
    while True:
        result = dls(start, depth, [start], {start})
        if result is not None:
            return result
        depth += 1


start = (5, 4, 0,
         6, 1, 8,
         7, 3, 2)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = ids(start, goal)

print("Solution:")
for state in solution:
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()
```

```python
# 8 Puzzle Problem using DFS

# Goal state
GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

def get_neighbors(state):
    """Generate possible moves."""

    neighbors = []

    zero_pos = state.index(0)
    row = zero_pos // 3
    col = zero_pos % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_pos = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with the selected tile
            new_state[zero_pos], new_state[new_pos] = \
                new_state[new_pos], new_state[zero_pos]

            neighbors.append(tuple(new_state))

    return neighbors


def dfs(start):
    """Depth First Search."""

    stack = [(start, [start])]
    visited = set()

    while stack:

        state, path = stack.pop()

        # If goal is reached
        if state == GOAL:
            return path

        if state in visited:
            continue

        visited.add(state)

        # Add neighboring states to stack
        for next_state in get_neighbors(state):

            if next_state not in visited:
                stack.append(
                    (next_state, path + [next_state])
                )

    return None


def print_puzzle(state):
    """Print puzzle in 3x3 format."""

    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])

    print()


# -------------------------
# Main Program
# -------------------------

start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

print("Initial State:")
print_puzzle(start)

solution = dfs(start)

if solution:

    print("Solution found!")
    print("Number of moves:", len(solution) - 1)

    print("\nSteps:")

    for i, state in enumerate(solution):

        print("Step", i)
        print_puzzle(state)

else:
    print("No solution found.")
```

```python
import heapq
# A* using Misplaced Tiles

initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Heuristic: Number of misplaced tiles
def heuristic(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Generate possible moves
def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


# A* algorithm
def astar():
    pq = []

    # (f, g, state)
    heapq.heappush(pq, (heuristic(initial), 0, initial))

    cost = {initial: 0}
    parent = {initial: None}

    while pq:
        f, g, current = heapq.heappop(pq)

        if current == goal:
            break

        for next_state in get_neighbors(current):

            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:

                cost[next_state] = new_g
                h = heuristic(next_state)
                new_f = new_g + h

                parent[next_state] = current
                heapq.heappush(pq, (new_f, new_g, next_state))

    # Construct path
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    print("Solution:")

    for state in path:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()

    print("Number of moves:", len(path) - 1)


astar()
```

```python
import heapq

initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Heuristic: Manhattan Distance
def heuristic(state):
    distance = 0

    for i in range(9):

        if state[i] == 0:
            continue

        # Current position
        row = i // 3
        col = i % 3

        # Goal position
        goal_pos = goal.index(state[i])
        goal_row = goal_pos // 3
        goal_col = goal_pos % 3

        distance += abs(row - goal_row)
        distance += abs(col - goal_col)

    return distance


# Generate possible moves
def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


# A* algorithm
def astar():
    pq = []

    heapq.heappush(pq, (heuristic(initial), 0, initial))

    cost = {initial: 0}
    parent = {initial: None}

    while pq:
        f, g, current = heapq.heappop(pq)

        if current == goal:
            break

        for next_state in get_neighbors(current):

            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:

                cost[next_state] = new_g

                h = heuristic(next_state)
                new_f = new_g + h

                parent[next_state] = current
                heapq.heappush(pq, (new_f, new_g, next_state))

    # Construct path
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    print("Solution:")

    for state in path:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()

    print("Number of moves:", len(path) - 1)


astar()
```
